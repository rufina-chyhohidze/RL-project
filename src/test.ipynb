### Effect of PPO Clip Range (Policy Update Constraint)

As an extension, we investigated the impact of PPO’s `clip_range` parameter in the custom-reward environment. The clip range limits how much the policy is allowed to change during each update and is known to strongly affect training stability.

Our results show a clear relationship between clip range size and performance. Smaller clip ranges lead to more stable learning and consistently successful policies. The best-performing configuration achieved an average return of **91.2** over 10 evaluation episodes, with all episodes successfully reaching the goal and very low variance. A medium clip range also achieved consistent success but with slightly lower average return (**89.1**).

In contrast, the largest clip range resulted in unstable learning, with one complete failure episode (return ≈ −99.9) and significantly higher variance, yielding a much lower average return (**50.1**).

These results demonstrate that, while reward shaping enables learning, careful tuning of PPO’s policy update constraints is essential for achieving both high performance and stability.
